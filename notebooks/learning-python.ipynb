### Loops

dont' forget that python is a whitespace/indent specific language

```python
print("For range Loop")
for i in range(10):
	print(i)
```

```python

print("For each Loop")
things = [1,2,3,4,5] # list
for thing in things:
	print(thing)
```

```python

print("While Loop")
count = 0

while (count < 10):
	print(count)
	count += 1
```
